```python
df_valid_claims = spark.sql("select * from non_vnet_ws.default.total_claim_amount")
display(df_valid_claims)
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import sum, col

Window_spec = Window.orderBy("total_claim_amount").rowsBetween(Window.unboundedPreceding, Window.currentRow)

df_running_total = df_valid_claims.withColumn("Running_total", sum(col("total_claim_amount")).over(Window_spec))
display(df_running_total)
```
